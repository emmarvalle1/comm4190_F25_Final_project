# <p style="color:blue">Project idea: RenovAI</p>

*<p style="color:purple">The idea pitch (short 100-150 word overview of the project idea):</p>*

RenovAI is an AI-powered renovation copilot that turns home design  inspiration into actionable renovation plans. Users upload photos of their space, provide rough dimensions, a budget, and a style goal (e.g. modern kitchen under $10,000). RenovAI analyzes the layout, identifies the correct materials, and generates feasible design options with cost estimates based on regional pricing. It assembles a curated shopping list and provides reviews of design choices; it also goes further to match users with local contractors when they are ready to execute. Unlike inspiration platforms (such as Pinterest) or contractor marketplaces (such as Thumbtack), RenovAI bridges the gap between the idea and execution. It provides personalized recommendations with specific dimensions, products, and cost, making users make informed decisions with confidence.

*<p style="color:purple">Description of the target users:</p>*

* First time homeowners who feel overwhelmed by renovation decisions
* Budget-conscious renovators seeking an affordable, guided approach
* Renters or students wanting reversible/low-commitment upgrades
* Landlord and property managers who need repeatable cost-efficient updates


*<p style="color:purple">Problems solved:</p>*

* Not knowing where to start
* Uncertainty around costs and feasibility 
* Difficulty translating aesthetics into real plans
* Distrust and overwhelm when choosing contractors 


*<p style="color:purple">More detailed summary of what the application will do:</p>*

* Users upload photos and provide dimensions and budget
* Computer vision identifies layout, fixtures, and materials
* An LLM interprets goals and produces a structured renovation brief
* The system generates 2-3 design plans with tradeoffs and rationale
* Cost forecasting engine uses regional labour and material pricing to estimate total cost
* A curated shopping list is provided with product recommendations
* Contractor matching ranks professionals based on specialization, pricing, and reviews
* Optional AR preview shows proposed designs in the user’s actual space
    * For example, you take a photo of your kitchen and the LLM overlays new cabinets, colors, counters, or lighting on top, so it looks like the renovation is already done.


*<p style="color:purple">What data (from users, web, other sources, etc) will be needed:</p>*

User data:
* Photos or videos of room
* Dimensions
* Budget
* Style preference 
* Constraints (e.g renter-friendly)
* Zip code for pricing and contractor search


External data:
* Retail product catalogues (prices, stock, spaces)
* Regional labor pricing data
* Contractor profiles (license/insurance status, ratings, job history)
* Basic building code and clearance standards 


App-generated data:
* Room measurement graph
* Bill of materials
* Cost predictions
* AR visualization assets
* Contractor brief documents


*<p style="color:purple">What is the role of LLMs:</p>*

The LLM: 
* Translates user goals into structured renovation plans
* Writes rationales, tradeoff explanations, timeline, and contractor briefs
* Summarizes product options and availability 
* Guides the user in back-and-forth conversation to refine design scope

The vision model handles spatial understanding and design; the cost engine handles pricing.

*<p style="color:purple">How your product will differ from both existing non-LLM powered systems in the domain and from just using a general purpose LLM for the task?</p>*

In comparison to non-LLM tools (Pinterest, Houzz, Angi), RenovAI provides personalized, scoped plans with costs and shopping lists.These platforms only provide inspiration or lists of contractors.

In comparison to general LLMs (ChatGPT, Gemini, Claude), RenovAI is grounded in real pricing data, real product availability, and contractor networks.  It is much more customizable and specialised in the renovations domain. It also produces actionable outputs (bill of quantities, AR previews, estimates) rather than generic suggestions.
