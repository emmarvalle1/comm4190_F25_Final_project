# **Router Workflow – Classifying User Intent**

RenovAI uses a lightweight router model to direct user queries into the correct workflow.

For my two scenarios, the router detects:

* **Kitchen-like renovation prompts → Scenario 1 Workflow**
* **Bathroom moisture, damage, or repair prompts → Scenario 2 Workflow**
* **Anything else → General RenovAI response**

# **STEP 4 — Implementation Section**

### **Implementation (Prompt-Based Approach)**

I implemented all workflows using **prompt chaining**, similar to my partner’s approach. No LangChain or code graph was required.

---

### **Router LLM**

Classifies messages into:

* **Scenario 1 (Kitchen)** — if photos or text mention kitchen, cabinets, counters, backsplash, etc.
* **Scenario 2 (Bathroom moisture)** — if moisture, stains, leaks, peeling, venting, or mold is detected
* **Other** — generic RenovAI assistant response

---

### **Scenario 1 Module – Kitchen Refresh**

Uses my **Kitchen Refresh structured prompt** from Milestone 3. Produces:

* Room/surface analysis
* Vision-based interpretation of layout
* **Three options (Basic, Mid, Premium)**
* Bill of Materials
* Estimated cost ranges
* Risks + constraints
* Next-step recommendations

This matches the workflow diagram above.

---

### **Scenario 2 Module – Bathroom Moisture Diagnostics**

Uses my **Bathroom Moisture structured prompt** from Milestone 3. Produces:

* Defects (stains, grout cracks)
* Risk scores (ventilation vs required airflow)
* Tiered repair plans
* Timelines and trade sequencing
* Budget estimate and safety notes

This also aligns exactly with the workflow diagram.
