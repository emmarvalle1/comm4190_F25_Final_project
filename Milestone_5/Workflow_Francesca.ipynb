## Router Workflow – Classifying User Intent

RenovAI uses a simple router model to determine which workflow should handle a user’s request.  
For my two scenarios, the router only needs to detect:

- **Kitchen-like renovation requests → Scenario 1 Workflow**  
- **Renter-safe bedroom or décor improvements → Scenario 2 Workflow**  
- **Other topics → generic RenovAI response**  




```mermaid
flowchart TD
    U[User Input<br/>text + optional images] --> R{Router LLM<br/>Identify Scenario Type}

    R -->|Kitchen-like input<br/>photos, cabinets, counters, appliances| K[Scenario 1:<br/>Kitchen Refresh Workflow]

    R -->|Renter-safe / bedroom / decor / storage| B[Scenario 2:<br/>Renter-Friendly Makeover Workflow]

    R -->|Other or unrelated| O[Out-of-Scope Reply]

    K --> OUT[Final Output to User]
    B --> OUT
    O --> OUT


### Scenario 1 Workflow – Galley Kitchen Refresh

```mermaid
flowchart TD
    A[User Input:\nKitchen photo + goals + budget + constraints] --> P[Planner LLM]

    P --> V[Vision Analysis\nIdentify materials, layout, issues]
    P --> Q[Clarification Questions\nif any missing measurements]

    V --> OPT[Generate 2-3 Renovation Options\nOption A / Option B / Option C]
    Q --> OPT

    OPT --> COST[Estimate Costs\nper option using regional pricing]

    COST --> RISKS[Flag Risks\ncondo rules, electrical, plumbing limits]

    RISKS --> OUT[Final Structured Report]


### Scenario 2 Workflow – Renter-Friendly Bedroom Makeover

```mermaid
flowchart TD
    A[User Input:\nBedroom photo + budget + renter restrictions] --> P[Planner LLM]

    P --> V[Vision Analysis\nIdentify surfaces, lighting, layout, storage gaps]

    P --> C[Check Constraints\nno drilling, no painting, removable only]

    V --> REC[Generate Renter-Safe Design Ideas\nlighting, textiles, storage]

    C --> REC

    REC --> SHOP[Create Shopping List\nwithin budget]

    SHOP --> OUT[Final Makeover Plan\nnext steps]




---

# **STEP 4 — Implementation Section**
 
## Implementation (Prompt-Based)

We implemented the workflows using prompt-chaining rather than coding:

**Router LLM**
- Classifies input into:
  - Scenario 1 (Kitchen)
  - Scenario 2 (Renter safety)
  - Other

**Scenario 1 Module**
- Uses the galley kitchen structured prompt
- Produces:
  - Photo Analysis
  - Options A/B/C
  - Cost breakdowns
  - Risks
  - Next steps

**Scenario 2 Module**
- Uses renter-safe bedroom prompt
- Produces:
  - Room Graph
  - Renter-safe improvements
  - Storage solutions
  - Lighting plan
  - Shopping list


## Testing Against My Two Milestone 3 Scenarios

| Scenario | Expected Path | Actual Path | Notes |
|---------|---------------|-------------|-------|
| Scenario 1: Kitchen | Scenario 1 workflow | Scenario 1 workflow | Correct: router classifies this as "kitchen/renovation" |
| Scenario 2: Renter bedroom | Scenario 2 workflow | Scenario 2 workflow | Correct: renter constraints and bedroom features lead to this path |

### Evidence
We ran each scenario through the workflow:

**Scenario 1 → Router → Kitchen Workflow → Output**
- Output included: material analysis, options A-C, cost ranges, constraints, risks  
- Fully aligned with workflow steps.

**Scenario 2 → Router → Renter Workflow → Output**
- Output included: renter-safe advice, lighting, storage, textiles, and a budgeted shopping list  
- Matches workflow steps exactly.

---


## Workflow Testing

I tested both workflows using my detailed Milestone 3 scenarios and verified that the router and workflows behaved as expected.

### Testing Table

| Scenario | Expected Workflow | Actual Workflow | Passed? | Notes |
|---------|-------------------|-----------------|---------|-------|
| Scenario 1 – Kitchen Refresh | Kitchen workflow | Kitchen workflow | ✔️ | Router correctly detected “kitchen” and passed input to the Scenario 1 flow. Output included options, cost ranges, risks, and next steps. |
| Scenario 2 – Renter Bedroom | Renter workflow | Renter workflow | ✔️ | Router correctly detected “renter restrictions/bedroom.” Output matched workflow steps, including renter-safe options and a shopping list. |

### Evidence Summary

**Scenario 1:**  
- Router classified intent as *kitchen renovation*  
- Workflow produced: room analysis → options → pricing → risks → next steps  
- Output followed expected structure from Milestone 3  

**Scenario 2:**  
- Router classified intent as *renter-safe bedroom makeover*  
- Workflow produced: layout assessment → constraint check → recommendations → shopping list  
- Output followed expected structure from Milestone 3  

### Overall Assessment
Both workflows behaved as designed and produced consistent, structured outputs aligned with Milestone 3.  
No unexpected routing or breakdowns occurred.

---

## Conclusion

Milestone 5 required designing full workflows for our LLM application and demonstrating that they work with real scenarios. For my two scenarios, I successfully:

- Built a functional router workflow  
- Created two complete scenario-specific workflows  
- Implemented each workflow using prompt chaining  
- Tested both scenarios end-to-end  
- Verified that the workflow structure matches the intended behavior  

These workflows clearly show the reasoning steps RenovAI should take and demonstrate how the LLM can reliably support renovation planning tasks. This completes Milestone 5 for my portion of the group project.
