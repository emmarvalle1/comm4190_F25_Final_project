**Emma’s Scenarios**

*Scenario #1:Accessible Bathroom Remodel for Aging in Place*

Description: A retired couple wants to update their small bathroom to make it safer and more accessible without exceeding a $15,000 budget. They need guidance on ADA-compliant design options and local contractor expertise.


The user, Maria (age 68), uploads photos and measurements of her 7’x9’ bathroom. She tells RenovAI her goals: “make it senior-friendly, safer, and easier to clean—keep it simple and classic.” She selects a $15,000 budget and notes that she prefers minimal downtime (one-week max installation).
RenovAI’s vision model detects current fixtures, identifies the bathtub-shower combo as a fall risk, and measures clearances between the toilet, sink, and tub. The LLM translates Maria’s goals into a renovation brief emphasizing accessibility, slip resistance, and grab-bar placement.
The system produces three scoped options:
Essential Safety Upgrade – swap tub for a low-threshold shower, add grab bars and non-slip flooring.
Comfort+ Upgrade – includes new lighting, touchless faucet, and ADA-height toilet.
Full Modern Accessibility – wall-mounted sink for wheelchair access, handheld showerhead, and automatic humidity sensor fan.
RenovAI also provides an annotated diagram highlighting clearances, door swing optimization, and fixture placement per ADA guidelines. It estimates total cost with contingency and filters contractors experienced in accessibility remodels.
Maria chooses the Comfort+ Upgrade plan, reviews the AR visualization, and generates a contractor brief including scope, budget, and a one-week schedule.
Step-by-Step Interaction Breakdown
User uploads bathroom photos and enters dimensions + budget.


Vision model identifies layout, tub/shower configuration, and lighting.


LLM asks clarifying questions about mobility and safety needs.


LLM creates structured renovation brief (goals + constraints).


Cost engine calculates total estimates based on regional labor + materials.


System presents 3 scoped options (with annotated visuals).


User selects plan and generates contractor-ready brief + shopping list.


AR module previews final layout and verifies fixture spacing.


Data Description
Data Needed
User data: images, dimensions, budget, safety/accessibility preferences, zip code.


External data: ADA guideline database, regional contractor directory, pricing data for fixtures and flooring, product catalogs (Home Depot, Lowe’s).
Data Created
Annotated room layout (with safety zones and clearance dimensions).


Bill of Quantities (materials, fixtures).


Cost forecast and labor schedule.


Contractor-ready renovation brief (PDF).


AR visualization assets (3D bathroom mock-up).


Evaluation
Success Indicators
Generated design meets ADA safety standards.


Cost estimate within ±15% of verified contractor quotes.


User feedback: feels safer and confident in contractor brief accuracy
Possible Complications
Ambiguous or incomplete photos (system requests retakes).


Older bathrooms may conceal water damage — flagged as “inspection recommended.”


If user mobility details are unclear, AI prompts for clarification (“Do you use a walker or mobility aid?”).



*Scenario #2: Small Patio Transformation for Entertaining”*

Description: A young professional wants to turn a narrow 6’x12’ apartment patio into a cozy outdoor entertaining space for under $2,500.
The user, Jake, uploads photos of his small city apartment patio—currently empty except for a bike rack and concrete flooring. He tells RenovAI: “I want to make this an inviting spot for 4 people to hang out, with lighting and seating, but I can’t make permanent changes since I rent.”
RenovAI analyzes the space, detects railing height and available floor area, and classifies the surface as concrete. The LLM interprets his intent as a reversible design for small-space entertaining.
The system generates three design options:
Budget Cozy Corner – foldable bistro set, string lights, and outdoor rug.
Lounge Deck – modular decking tiles, sectional seating, and solar lanterns.
Entertainment Setup – weatherproof Bluetooth speakers, bar cart, and potted plant dividers for privacy.
Each design includes a shopping list with current prices and stock availability. The app provides an AR overlay so Jake can preview lighting and seating arrangements at night. RenovAI also generates an “assembly order” timeline and safety notes for balcony load limits.
Jake chooses the Lounge Deck plan, substitutes one item to fit budget, and saves the full plan to his project board.
Step-By-Step Interaction Breakdown
User uploads patio photos and dimensions, checks “renter-friendly” mode.


Vision model analyzes floor area, wall/railing boundaries, and lighting conditions.


LLM identifies key goals (entertaining, budget, reversible).


AI generates three design plans with tradeoffs.


Cost calculator ensures total stays under $2,500.


User refines design interactively (“add more greenery”).


AR view previews layout at night/day.


System outputs final shopping list + setup instructions.


Data Description
Data Needed
User data: images, dimensions, budget, renter restriction toggle.
External data: product catalogs (IKEA, Amazon, Wayfair), weatherproof material specs, regional shipping data.
Data Created
Layout and zoning plan (furniture, lighting).
Product list with prices and availability.
AR mock-up (day/night modes).
Setup checklist and budget summary.


Evaluation
Success Indicators
Total cost ≤ $2,500.
User satisfaction with AR visualization and product list.
Design respects renter constraints (no drilling or permanent fixtures).
Possible Complications
Limited space misinterpreted from skewed photos — app requests re-capture.
Inconsistent outdoor lighting data — system adjusts for estimated brightness.
Product out-of-stock — AI substitutes similar item automatically.
