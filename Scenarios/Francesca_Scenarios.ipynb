**Francesca’s scenarios**


*Scenario 1 - $10,000 Galley Kitchen Refresh in a 1980s Condo*


Short description: 
A first-time homeowner wants to refresh a small galley kitchen without moving plumbing or changing the layout. The priority is creating a bright and modern look while spending a maximum of $10,000. 


Narrative:
Francesca uploads 4 photos of her kitchen (2 wide shots, 1 of each counter run, and 1 showing lighting). She enters rough dimensions (10’ x 8’), and notes that the building is an older 1960s condo, setting a strict $10,000 total budget. She tells RenovAI she wants a “modern, bright, minimalist style” but cannot move plumbing due to building restrictions.


RenovAI’s computer vision identifies existing cabinet layout, flooring type, countertop material, and appliance placement. The system recognises limited natural light and outdated cabinet finishes. The LLM listens to her goals and generates a structured renovation brief, including constraints (no wall removal, no new electrical circuits). 


RenovAI provides 3 plan variations:
* “Touchpoint Refresh” - repaint cabinets, replace hardware, swap lighting, add peel-and-stick backsplash.
* “Mid-Range Update” - replace cabinet doors, install new countertops, add under-cabinet lighting.
* “High-Impact Swap” - open shelving on one side and replace sink and faucet set.

Each option includes a Bill of Quantities which lists materials and labor, as well as regionalized cost forecasts with contingency ranges. RenovAI checks inventory and only selects items in stock at local retailers to avoid delays.

Say Francesca picks the “Mid-Range Update”, RenovAI generates:
* A shopping list with links and price breakdowns
* An AR preview showing new cabinet finishes and countertop color
* A contractor brief describing scope, measurements, and required steps
* A timeline with estimated labor hours and order sequencing


Step-by-Step Interaction Breakdown:
* User uploads photos and dimensions
* Vision model detects layout, materials, and fixtures
* LLM converts goals into structured renovation scope
* Cost engine calculates regional pricing estimates
* System presents 3 scoped plans and the rationales
* User selects and refines plan
* System generates contractor-ready documents and product cart


Data Needed:
* Room photos, dimensions, budget, style moodwords, zip code
* Retailer product catalogs
* Regional labor cost database


Data Created:
* Room graph (measurements and object placement)
* Bill of Quantities
* Cost estimate model outputs
* Contractor brief PDF
* AR visualization assets


Evaluation and Potential Issues:

Success is measured by:
* The cost estimate being within ±20% of real contractor quotes
* The user understanding scope clearly without confusion
* The contractor brief being complete enough to request accurate bids


Possible complications include outdated electrical or hidden wall damage. RenovAI flags these as contingency risks and advises optional pre-inspection.


*Scenario 2 - Renter-friendly makeover under $800*

Short description:
A renter wants to significantly improve bedroom comfort, lighting, and storage without making irreversible changes.

Narrative:
Francesca uploads 3 photos of her bedroom and selects a maximum budget of $800. She also checks a box indicating the project must be renter-friendly: no repainting walls, no drilling into studs, and all upgrades must be removable.
RenovAI detects wall color, window orientation, current lighting quality, and available floor space. The LLM interprets her priorities: improved coziness, better sleep lighting, and additional storage.

RenovAI proposes a layered plan:
* Peel-and-stick accent wall behind bed
* Blackout curtains sized to window width (auto-measured via photo)
* Bedside task lighting to reduce blue light exposure
* Under-bed rolling storage drawers
* Adhesive cable channels to hide cords
* A “one-day install schedule” with task ordering
* The system generates a cart of reversible and low-cost products from IKEA, Wayfair, and Amazon, confirming they are all in stock. The app provides instructions for safe installation and how to remove everything without wall damage at the end of the lease.

To help the user feel confident, RenovAI provides:
* A before-and-after AR mock-up
* A step-by-step assembly & placement checklist
* A time breakdown (estimate: 3 hours total install)
* Throughout the process, the app keeps a running total so the user can swap items to stay under budget.

Step-by-Step Interaction Breakdown:
* User uploads photos, chooses renter-friendly mode
* System identifies reversible upgrades only
* Cost calculator tracks cart total dynamically
* AR preview displays final visual layout
* User purchases items and receives shipping updates
* App sends install reminders and tracks completion

Data Needed:
* Photos, budget, renter restriction toggle
* Retail product information
* Delivery estimates

Data Created:
* Shopping cart with alternatives for each item
* Install and removal checklist
* Before/after visualization


Evaluation:

Success is measured by:
* Budget ≤ $800
* Improved lighting and storage based on user satisfaction survey
* The user feeling confident they can undo everything without damage

Possible challenges: small rooms with limited floor storage; RenovAI responds by suggesting vertical storage or foldable options.
