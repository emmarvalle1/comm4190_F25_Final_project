
## Scenario 1 — “Snap → Plan → Hire” (Kitchen Refresh)
Short description
User snaps photos of their kitchen, enters measurements, and receives an AI-generated design plan with a shoppable bill of materials (BOM) and matched contractors.
Narrative
A homeowner wants to refresh their kitchen (new backsplash, counters, pulls) without a full gut. They upload 3–5 photos, add basic measurements, set a budget, and state a style (“modern, light woods, matte black hardware”). The app analyzes the images and dimensions, proposes layouts/finishes, estimates cost by region, and returns a curated product list plus 3 vetted local contractors. The user locks a plan, books a contractor, and exports a BOM.
Step-by-step interaction
User opens app → selects “Kitchen Refresh”.


User uploads images (wide, sink wall, stove wall) and enters measurements (L×W, counter run lengths, backsplash height).


App runs vision analysis (detects cabinets, appliances, materials) and geometry fit from measurements.


User enters budget and style (with optional inspo links).


AI generates 3 design options with mood boards + price bands.


User selects Option B → reviews BOM (tile, grout, pulls, faucet) and swaps a couple of SKUs.


Cost engine recalculates total (materials + labor, regional multipliers).


App suggests 3 contractors (tile specialist, kitchen installer, handyman) with availability.


User requests quotes; contractors get structured scope + photos + measurements.


User confirms one contractor, pays a deposit escrow; app locks the BOM and generates a project schedule.


### Data description
Needed (to run scenario):
Images (JPEG/HEIC) with EXIF; optional LiDAR depth if available.


Structured measurements (JSON): wall lengths, ceiling height, counter runs; entry via form or room-scan.


Location (ZIP) for pricing & code hints.


Preferences: style taxonomy, color palette, finish constraints; budget cap.


Catalog access: product feed with specs (dimensions, finish, price, lead times).


Contractor graph: specialties, license/insurance, ZIP coverage, historical NPS, availability.



Created (as a result):
BOM JSON (SKU, qty, unit price, lead time, alternates).


Design rationale (compact natural language + citations to specs).


Cost model snapshot (materials, labor, contingencies, region).


Contractor shortlist with structured scope + attachments.


Audit log (decisions, prompts, model versioning).


Project schedule (milestones, dependencies).




### Evaluation
Success criteria
Vision correctly identifies ≥95% of major surfaces (cabinets, counters, backsplash).


BOM dimensional fit error < 3%.


Cost estimate within ±12% of accepted contractor quote.


User books a contractor or exports a finalized BOM in ≤15 minutes median.


Test methods
Synthetic kitchens with known ground-truth dimensions.


A/B test design option ordering; measure conversion to quote request.


Back-test pricing vs. historical job outcomes in same ZIPs.


Complications & mitigations
Low-light or wide-angle distortion: prompt user to retake; apply lens correction; show uncertainty flags.


Measurement mismatch: run a “sanity check” vs. vision inference; if variance >8%, trigger a guided measurement wizard.


Unavailable SKUs / long lead times: auto-substitute with constraint satisfaction (match color/finish/price).


Adversarial inputs (unsafe/illegal asks): building-code ruleset gate; refuse unsafe designs and offer compliant alternatives.


Contractor fraud/quality drift: require license/insurance verification, escrow, milestone payments, and post-job QA surveys.


Model hallucination: ground all recommendations in product specs; show citations and allow “show sources”.



## Scenario 2 — “Diagnose → Compare → Optimize” (Bathroom Remodel with Moisture Issue)
### Short description
User uploads images of a bathroom with moisture spots, provides dimensions, and compares three remodel paths (minimal repair, mid-range refresh, premium). The app diagnoses likely causes, estimates costs, and books the right pro.
Narrative
A homeowner notices ceiling discoloration and peeling paint in the bathroom. They upload photos of the ceiling and shower area, enter room size and ventilation details, and select priorities (“fix moisture first, keep costs low”). The app detects probable moisture intrusion, proposes fix tiers (vent upgrade + repaint; partial tile + membrane; full shower rebuild), models cost/risk tradeoffs, and matches a waterproofing contractor and an electrician. The user chooses the mid-range plan, orders materials, and schedules trades in sequence.
Step-by-step interaction
User selects “Bathroom Troubleshoot” → uploads 4 photos; enters L×W×H, vent CFM (if known), and building type (condo).


App runs defect detection (stains, grout failure, caulk gaps) and risk model (likelihood of hidden moisture).


User picks constraints (max $7,500; 2-week window).


AI generates 3 paths: Minimal / Mid / Premium, each with scope, BOM, timeline, risk score.


User expands Mid: adds anti-fog mirror; removes niche shelf.


Cost engine recalculates; timeline reflows (tile cure times).


App recommends two trades: waterproofing tile specialist + electrician; proposes sequencing and availability.


User books site visit; app exports inspection checklist and AR overlay to mark demo lines on wall.


After inspection, estimates adjust; user confirms and places orders for moisture-rated drywall + shower system.


App tracks delivery, milestones, and change orders.


### Data description
Needed:
Photos (including close-ups of stains/tiles), dimensions, vent info (CFM or “unknown”), building type (house/condo), budget/time constraints.


Product catalogs for moisture-rated materials (cement board, membranes, anti-mold paint).


Trades directory with sub-specialties (waterproofing, ventilation/electrical), compliance requirements for condos (board approval, quiet hours).


Provided to LLM/stack:
Defect model outputs a defect_map with labels (stain_class, grout_crack, mold_prob) + bounding boxes + confidence.


Option engine prompt includes constraints and code hints (e.g., required vent CFM per room size).


Created:
Tiered scopes with tasks and dependencies.


Risk register (likelihood × impact; e.g., hidden substrate rot).


AR markup package (planes, demo lines, “no-cut” zones).


Trades sequence (visit → demo → waterproof → tile → electrical → paint).


Compliance packet (exportable PDF checklist for condo board if selected).



### Evaluation
Success criteria
Defect classifier AUROC ≥ 0.92 on curated bathroom dataset.


Proposed vent CFM meets or exceeds local code baseline for room volume.


Risk-adjusted cost/timeline within ±15% post-inspection.


User completes quotes → books trades in ≤2 sessions.


Test methods
Golden set of annotated bathroom issues; inter-rater agreement ≥0.85.


Simulated schedules with random delays; verify dependency handling.


Condo compliance scenarios with required documentation exports.


Complications & mitigations
Hidden damage not visible in photos: communicate uncertainty; require inspection checkpoint before final pricing.


Condos/HOAs approvals: auto-generate compliance packet; surface quiet hours & debris rules.


Moisture/mold safety: display safety advisory; recommend certified remediation when thresholds exceed.


User inputs “unknown” measurements: trigger scan-assist and conservative BOM with adjustable buffers.


Conflicting user goals (cheap + premium finishes): show trade-off visualizer and let user pin constraints.


Privacy: blur faces/personal items; on-device pre-processing for PII; clear data retention settings.
