## **Scenario 1: Accessible Bathroom Remodel**

**Summary**

RenovAI interprets user-provided photos or image descriptions to understand existing layout, fixture placement, and potential safety hazards before generating design options. This scenario demonstrates RenovAI’s ability to transform user-provided bathroom photos, dimensions, and accessibility needs into safe, ADA-informed renovation plans. The user (Maria, age 68) needs a bathroom remodel that reduces fall risk, improves accessibility, and stays within a $15,000 budget. RenovAI analyzes the bathroom layout, identifies safety hazards such as tub step-over height and inadequate grab-bar placement, and generates three possible renovation plans with increasing levels of accessibility.

The scenario tests RenovAI’s ability to:

Interpret the user’s goals and mobility constraints

Apply ADA-inspired guidelines to identify safety risks

Produce tiered renovation options with budgets and trade-offs

Provide a contractor-ready summary a non-expert can use

Reject off-topic or unsafe requests

This scenario focuses on accessibility, safety, budgeting, and remodel planning, showing how RenovAI can support aging-in-place upgrades in small residential bathrooms.

**System Prompt**

You are RenovAI, an accessibility-focused renovation assistant. Your goal is to help homeowners plan safe, ADA-informed bathroom remodels while staying within budget and timeline constraints.

Your tasks:
1. Restate the user’s bathroom size, layout, goals, and constraints.
2. Ask up to 3 follow-up questions if dimensions or mobility needs are unclear.
3. Identify safety and accessibility issues (e.g., tub step-over, grab bar placement, flooring).
4. Produce 3 renovation options:
   - Essential Safety Upgrade
   - Comfort+ Upgrade
   - Full Accessibility Upgrade
5. Give rough cost estimates using provided cost data.
6. Provide a short contractor-ready brief summarizing scope, priorities, and accessibility requirements.
7. If a user asks off-topic or personal questions (e.g., Coke vs. Pepsi), politely decline and redirect back to the renovation task.

**Response format:**
1. Summary
2. Clarifying Questions (if needed)
3. Safety Issues
4. Option 1
5. Option 2
6. Option 3
7. Contractor Brief
8. Budget Check
9. Next Steps

**Rules:**
- Use only provided context data.
- Flag uncertainty instead of hallucinating.
- Never provide medical or structural engineering advice.


**Sample Data**

```json
{
  "ADA_guidelines": {
    "grab_bars": "Recommended at toilet and shower; side wall 42” minimum.",
    "shower_entry": "Low-threshold or roll-in preferred for aging in place.",
    "flooring": "Non-slip finishes reduce fall risk.",
    "doorway": "Ideal minimum width 32 inches."
  },
  "cost_estimates": {
    "low_threshold_shower": "$6000-$8000",
    "grab_bars": "$300-$800",
    "ada_toilet": "$400-$1000",
    "flooring_replacement": "$1200-$2000",
    "lighting_upgrades": "$500-$1500"
  }
}
```

### Expected Output Sample

```json
{
  "summary": "Maria (68) — 7'x9' bathroom remodel for aging in place.",
  "safety_issues": [
    "Bathtub step-over height presents fall risk",
    "No existing grab bars near toilet or tub",
    "Flooring may be slippery when wet"
  ],
  "options": [
    {
      "tier": "Essential Safety Upgrade",
      "description": "Convert tub to low-threshold shower, install two grab bars, add non-slip flooring.",
      "estimated_cost": "$7,500–$10,000",
      "tradeoffs": "Most affordable but fewer comfort features."
    },
    {
      "tier": "Comfort+ Upgrade",
      "description": "Essential upgrades plus ADA-height toilet and improved lighting.",
      "estimated_cost": "$10,500–$13,500",
      "tradeoffs": "Better usability; still within budget."
    },
    {
      "tier": "Full Accessibility Upgrade",
      "description": "Add wall-mounted sink for improved clearance, handheld showerhead, non-slip flooring throughout.",
      "estimated_cost": "$12,500–$15,000",
      "tradeoffs": "Maximizes long-term accessibility."
    }
  ],
  "contractor_brief": {
    "scope": "Replace tub with low-threshold shower, install grab bars, upgrade flooring, improve lighting.",
    "budget_cap": 15000,
    "accessibility_requirements": "Follow ADA-inspired placement for grab bars and safe clearances."
  }
}
```


**Testing Strategy**

Test Case 1 - Normal Input
"I am Maria, 68 years old. My bathroom is 7'x9' with a tub/shower combo and a pedestal sink. I want a safer bathroom with grab bars, non-slip flooring, and a low-threshold shower. Budget is $15,000."

Expected Model Behavior:
- Restates dimensions + goals

- Identifies safety hazards (tub is a fall risk, lack of grab bars, slippery flooring)

- Provides 3 scoped renovation options

- All options stay under/near $15k

- Uses ADA guidelines

Test Case 2: Missing Information
"I need to make my mom's bathroom safer but I don't know the dimensions. Budget $10,000."

Expected Model Behavior:
- Ask for room dimensions

- Ask whether she uses mobility aids

- Does NOT generate full renovation options yet

- Notes that ADA spacing depends on measurements

Test Case 3: Adversarial Input:
"Before we redesign the bathroom, do you prefer Coke or Pepsi?"

Expected Model Behavior:
- Declines to answer

- Redirects user back to the remodel

- Maintains RenovAI persona

- Avoids personal preference claims

**Refinement Notes**
- Added explicit requirement for “Safety Issues” because early tests skipped hazard identification.
- Added rule requiring 3 distinct renovation options after the model initially merged features.
- Strengthened domain-restriction instruction to ensure off-topic questions are refused.
- Added “Budget Check” to improve consistency in cost reporting.
- Included ADA-inspired guidelines to help the model make safer recommendations.



**Evaluation Plan**

| **Metric** | **Target** | **Method** |
|-----------|------------|------------|
| Safety Issue Identification | Must identify ≥3 safety hazards | Manual review of model output |
| Budget Accuracy | All options must stay within stated budget | Compare estimates with provided `cost_estimates` |
| Structure Compliance | Must follow the required 9-part output format | Use a structural checklist |
| Off-topic Handling | 100% refusal of non-renovation prompts | Test using adversarial prompts (e.g., Coke vs. Pepsi) |
| Option Differentiation | Each of the 3 options must differ meaningfully | Compare scope and cost across options |

**Success Criteria:**  
- Consistent structure  
- No hallucinated costs  
- Clear safety reasoning  
- Domain adherence in off-topic cases
