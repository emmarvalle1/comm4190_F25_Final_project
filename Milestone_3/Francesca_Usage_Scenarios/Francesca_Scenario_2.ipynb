## 6. Removal Guide (Lease-End Safe)

### Wall Treatments
1. **Peel-and-Stick Wallpaper:** Slowly pull at 45° angle from top corner. Heat stubborn edges with hairdryer (low setting) for 10 seconds. Wipe residue with damp cloth + drop of dish soap.
2. **Command Strips (hooks, shelf, tapestry, frames):** Pull tab straight down parallel to wall (do NOT pull away from wall). Hold for 30 sec. Strips stretch and release without damage.
3. **LED Strip Lights:** Gently peel adhesive backing. Remove residue with rubbing alcohol on microfiber cloth.

### Window Treatments
- Remove curtains from tension rod
- Twist tension rod counterclockwise to release pressure
- No marks left on window frame

### Flooring
- Roll up area rug (vacuum underneath first)
- Carpet beneath will be undisturbed

### Furniture & Storage
- Disassemble metal rack (no tools needed, pieces separate)
- Roll garment rack to new location
- Slide under-bed bins out
- Remove over-door organizer (hangs on hooks, zero damage)

### Lighting
- Unplug all devices (smart outlets, lamps, diffuser, sound machine)
- Remove adhesive puck lights (peel gently, clean with alcohol)
- Coil and store LED strips

### Textiles
- Wash/pack throw blanket, pillow covers, table runners
- Vacuum-seal if storing

### Small Items
- Remove drawer dividers
- Detach bedside caddy
- Peel cable management clips (rubbing alcohol for residue)

**DAMAGE RISK ASSESSMENT:**  
✅ **Zero permanent damage** when removal instructions followed  
⚠️ **Minor risk:** Command strips on textured walls (test spot first)  
⚠️ **Minor risk:** Peel-and-stick wallpaper on old paint (edges may lift paint if >10 years old — prime first if concerned)

**DEPOSIT PROTECTION:**  
- Photograph walls before applying any adhesive products
- Save all product packaging showing "removable" or "damage-free" claims
- Remove items 48 hours before move-out inspection to allow time for residue cleanup
